```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

students = []
for student_id in range(1, 301):

    grade = rng.choice(["g1", "g2", "g3"])

    if grade == "g1":
        math = round(rng.normal(loc=53, scale=19))
        english = round(rng.normal(loc=55, scale=19))
        japanese = round(english + rng.normal(loc=0, scale=10))
        social = round(rng.normal(loc=55, scale=19))
        science = round(math + rng.normal(loc=0, scale=10))
        
    elif grade == "g2":
        math = round(rng.normal(loc=50, scale=18))
        english = round(rng.normal(loc=58, scale=18))
        japanese = round(english + rng.normal(loc=0, scale=9))
        social = round(rng.normal(loc=58, scale=18))
        science = round(math + rng.normal(loc=0, scale=9))
        
    else: #g3
        math = round(rng.normal(loc=55, scale=15))
        english = round(rng.normal(loc=68, scale=15))
        japanese = round(english + rng.normal(loc=0, scale=8))
        social = round(rng.normal(loc=61, scale=15))
        science = round(math + rng.normal(loc=0, scale=8))

    math = max(0, min(100, math))
    english = max(0, min(100, english))
    japanese = max(0, min(100, japanese))
    social = max(0, min(100, social))
    science = max(0, min(100, science))
    
    average_score = (
        math + english + japanese + science + social
    ) / 5
    
    attendance = round(
        60 + average_score * 0.3 + rng.normal(loc=0, scale=8)
    )
    
    attendance = max(0, min(100,attendance))
    
    students.append({
        "student_id":student_id,
        "grade":grade,
        "math":math,
        "english":english,
        "japanese":japanese,
        "social":social,
        "science":science,
        "attendance":attendance,
    })

df = pd.DataFrame(students)
```

```python
df["average_score"] = (
    df["math"]
    + df["english"]
    + df["japanese"]
    + df["science"]
    + df["social"]
) / 5
```

```python
columns = [
    "student_id",
    "grade",
    "math",
    "english",
    "japanese",
    "social",
    "science",
    "attendance"
]

df[columns].to_csv("../data/student_data.csv", index=False)
```
